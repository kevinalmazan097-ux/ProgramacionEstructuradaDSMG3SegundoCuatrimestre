# Programa de calificaciones 
```
este programa calcula el promedio de tres calificaciones y evalua si un alumno aprovo  o  no  aprovo 

```
## Definicion de Variables


````python
calif1 = 0.0
calif2 = 0.0
calif3 = 0.0
promedio = 0.0

````

## Entrada de Datos

````python
calif1 = float(input("Introduce la Calificacion1"))
calif2 = float(input("Introduce la Calificacion2"))
calif3 = float(input("Introduce la Calificacion3"))
````

## Proceso: Calculo de Promedio

````python
promedio=(calif1+calif2+calif3)/3
````

## Comparacion con una Estructura selectiva double

````python
if promedio >= 7.0:
    print("Aprovado")
else:
    print("No Aprovado")
    
````
